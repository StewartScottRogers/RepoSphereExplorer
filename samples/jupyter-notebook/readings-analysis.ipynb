# Weather readings, September 2026

Loads `samples/csv/readings.csv`, checks it for gaps, and plots the
daily range per station. Every cell below has been run, so the stored
outputs are what a reader sees before executing anything.

```python
import csv
import statistics
from collections import defaultdict
from pathlib import Path

READINGS = Path("../csv/readings.csv")
```

```python
with READINGS.open(newline="", encoding="utf-8") as handle:
    rows = list(csv.DictReader(handle))

stations = {row["station_id"] for row in rows}
print(f"{len(rows)} rows across {len(stations)} stations")
```

## Daily range

The range is the difference between the warmest and coolest reading of
the day. A station with fewer than three readings is reported rather
than silently averaged.

```python
by_station = defaultdict(list)
for row in rows:
    by_station[row["station_id"]].append(float(row["celsius"]))

ranges = {
    station: round(max(values) - min(values), 1)
    for station, values in by_station.items()
    if len(values) >= 3
}
sorted(ranges.items(), key=lambda pair: -pair[1])[:5]
```
